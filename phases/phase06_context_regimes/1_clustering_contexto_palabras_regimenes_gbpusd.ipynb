
# Phase 6 - Clustering de Contexto (Palabras/Regimenes) sobre GBPUSD

Objetivo: formar "palabras/regimenes" de contexto sobre la secuencia de regímenes del mejor modelo (`v5.2`),
con validación temporal walk-forward y evaluación OOS.

Entrada:
- `phase5/output_superclustering_contextual_GBPUSD_v5_2_chronos_pretrained/GBPUSD_context_window_embeddings_v5_2.parquet`

Salida:
- Asignación de `word_id` por ventana contextual
- Perfil económico por `word_id`
- Matriz de transición entre palabras
- Resumen de métricas OOS


```python

import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD, PCA
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score, adjusted_rand_score
from scipy.stats import kruskal, f_oneway

pd.set_option('display.max_columns', 260)
pd.set_option('display.width', 340)

# =========================
# Config
# =========================
SEED = 42
SYMBOL = 'GBPUSD'
CONTEXT_LEN = 24
NGRAM_RANGE = (2, 4)
MIN_DF = 3
MAX_DF = 0.95
SVD_DIM = 64

K_CANDIDATES = list(range(4, 13))
KMEANS_SEEDS = [11, 42, 77]
N_INIT_KMEANS = 50
MIN_CLUSTER_N_FOLD = 15
MIN_CLUSTER_N_TEST = 20

# walk-forward folds sobre train+val
FOLD_SPECS = [
    (0.50, 0.10),
    (0.60, 0.10),
    (0.70, 0.10),
    (0.80, 0.10),
]

ROOT = Path('.')
IN_PQ = ROOT / 'phases/phase05_context_sequencing/output_superclustering_contextual_GBPUSD_v5_2_chronos_pretrained/GBPUSD_context_window_embeddings_v5_2.parquet'
IN_SUMMARY = ROOT / 'phases/phase05_context_sequencing/output_superclustering_contextual_GBPUSD_v5_2_chronos_pretrained/GBPUSD_superclustering_summary_v5_2.json'

OUT_DIR = ROOT / 'phases/phase06_context_regimes/output_context_words_GBPUSD_v6'
OUT_DIR.mkdir(parents=True, exist_ok=True)

WORDS_PQ = OUT_DIR / 'GBPUSD_context_words_windows_v6.parquet'
WORDS_CSV = OUT_DIR / 'GBPUSD_context_words_windows_v6.csv'
PROFILE_CSV = OUT_DIR / 'GBPUSD_context_words_profile_v6.csv'
TRANS_CSV = OUT_DIR / 'GBPUSD_context_words_transition_matrix_v6.csv'
MODEL_SEL_CSV = OUT_DIR / 'GBPUSD_context_words_k_selection_v6.csv'
PROTOTYPES_CSV = OUT_DIR / 'GBPUSD_context_words_prototypes_v6.csv'
SUMMARY_JSON = OUT_DIR / 'GBPUSD_context_words_summary_v6.json'

np.random.seed(SEED)


def zcol(s: pd.Series):
    s = pd.to_numeric(s, errors='coerce')
    m = float(s.mean())
    sd = float(s.std(ddof=0))
    if not np.isfinite(sd) or sd < 1e-9:
        return pd.Series(np.zeros(len(s)), index=s.index, dtype=float)
    return (s - m) / (sd + 1e-9)


def build_folds(idx_tv: np.ndarray):
    n = len(idx_tv)
    folds = []
    for tr_r, va_r in FOLD_SPECS:
        tr_end = int(n * tr_r)
        va_end = min(n, int(n * (tr_r + va_r)))
        if va_end - tr_end < 30:
            continue
        folds.append((idx_tv[:tr_end], idx_tv[tr_end:va_end]))
    return folds


# =========================
# Load base stream
# =========================
base = pd.read_parquet(IN_PQ).copy()
base['start_ts'] = pd.to_datetime(base['start_ts'], errors='coerce')
base['end_ts'] = pd.to_datetime(base['end_ts'], errors='coerce')
base = base.sort_values(['end_ts', 'window_id']).reset_index(drop=True)

required = ['window_id', 'start_ts', 'end_ts', 'split', 'regime_id', 'regime_token', 'ret_fwd_24h_pips']
missing = [c for c in required if c not in base.columns]
if missing:
    raise ValueError(f'Missing required columns in input: {missing}')

print('base windows =', len(base))
print('base split counts:', base['split'].value_counts().to_dict())

# =========================
# Build contextual windows over regime tokens
# =========================
reg_tokens = base['regime_token'].astype(str).to_numpy()
reg_ids = base['regime_id'].astype(int).to_numpy()
splits = base['split'].astype(str).to_numpy()
rets = pd.to_numeric(base['ret_fwd_24h_pips'], errors='coerce').to_numpy(dtype=float)
starts = base['start_ts'].to_numpy(dtype='datetime64[ns]')
ends = base['end_ts'].to_numpy(dtype='datetime64[ns]')
window_ids = base['window_id'].to_numpy(dtype=int)

rows = []
for i in range(CONTEXT_LEN - 1, len(base)):
    j0 = i - CONTEXT_LEN + 1
    seq = reg_tokens[j0:i+1]
    rows.append({
        'ctx_window_id': int(i - (CONTEXT_LEN - 1)),
        'base_window_id_end': int(window_ids[i]),
        'start_ts': pd.Timestamp(starts[j0]),
        'end_ts': pd.Timestamp(ends[i]),
        'split': str(splits[i]),
        'token_seq': seq.tolist(),
        'token_text': ' '.join(seq.tolist()),
        'last_regime_id': int(reg_ids[i]),
        'ret_fwd_24h_pips': float(rets[i]) if np.isfinite(rets[i]) else np.nan,
    })

CTX = pd.DataFrame(rows)
print('context windows =', len(CTX))
print('context split counts:', CTX['split'].value_counts().to_dict())

idx_tr = np.where(CTX['split'].to_numpy() == 'train')[0]
idx_va = np.where(CTX['split'].to_numpy() == 'val')[0]
idx_te = np.where(CTX['split'].to_numpy() == 'test')[0]
idx_tv = np.concatenate([idx_tr, idx_va])

ret_ctx = pd.to_numeric(CTX['ret_fwd_24h_pips'], errors='coerce').to_numpy(dtype=float)
texts = CTX['token_text'].astype(str).to_numpy()

# =========================
# Walk-forward K selection
# =========================
folds = build_folds(idx_tv)
print('folds (train,val sizes):', [(len(a), len(b)) for a, b in folds])

# Precompute embeddings per fold (fit vectorizer/svd/scaler only on fold train)
fold_repr = []
for f_id, (tr_idx, va_idx) in enumerate(folds):
    vec = TfidfVectorizer(
        analyzer='word',
        lowercase=False,
        token_pattern=r'[^\s]+',
        ngram_range=NGRAM_RANGE,
        min_df=MIN_DF,
        max_df=MAX_DF,
    )
    Xtr_sp = vec.fit_transform(texts[tr_idx])
    Xva_sp = vec.transform(texts[va_idx])

    n_feat = Xtr_sp.shape[1]
    n_comp = int(min(SVD_DIM, max(2, n_feat - 1)))

    if n_feat <= 2:
        # fallback extremely low-vocab case
        Ztr = Xtr_sp.toarray()
        Zva = Xva_sp.toarray()
    else:
        svd = TruncatedSVD(n_components=n_comp, random_state=SEED)
        Ztr = svd.fit_transform(Xtr_sp)
        Zva = svd.transform(Xva_sp)

    sc = StandardScaler()
    Ztr = sc.fit_transform(Ztr)
    Zva = sc.transform(Zva)

    fold_repr.append((tr_idx, va_idx, Ztr, Zva))

rows_sel = []
for k in K_CANDIDATES:
    fold_eta2, fold_kw = [], []
    fold_sil, fold_ch, fold_db, fold_ari = [], [], [], []

    for tr_idx, va_idx, Ztr, Zva in fold_repr:
        seed_labels = {}
        sils, chs, dbs = [], [], []

        for sd in KMEANS_SEEDS:
            km = KMeans(n_clusters=k, n_init=N_INIT_KMEANS, random_state=sd)
            km.fit(Ztr)
            lv = km.predict(Zva)
            seed_labels[sd] = lv
            if len(np.unique(lv)) >= 2 and len(lv) > k:
                sils.append(float(silhouette_score(Zva, lv)))
                chs.append(float(calinski_harabasz_score(Zva, lv)))
                dbs.append(float(davies_bouldin_score(Zva, lv)))

        aris = []
        keys = list(seed_labels.keys())
        for i in range(len(keys)):
            for j in range(i + 1, len(keys)):
                aris.append(float(adjusted_rand_score(seed_labels[keys[i]], seed_labels[keys[j]])))

        lv = seed_labels[42]
        rv = ret_ctx[va_idx]
        ok = np.isfinite(rv)
        rv = rv[ok]
        lv = lv[ok]

        groups = []
        for cid in np.unique(lv):
            g = rv[lv == cid]
            if len(g) >= MIN_CLUSTER_N_FOLD:
                groups.append(g)

        if len(groups) >= 2:
            kw = float(kruskal(*groups).pvalue)
            all_vals = np.concatenate(groups)
            gm = np.mean(all_vals)
            ssb = sum(len(g) * (np.mean(g) - gm) ** 2 for g in groups)
            sst = np.sum((all_vals - gm) ** 2)
            eta2 = float(ssb / sst) if sst > 0 else np.nan
            fold_kw.append(kw)
            fold_eta2.append(eta2)

        if sils:
            fold_sil.append(float(np.mean(sils)))
        if chs:
            fold_ch.append(float(np.mean(chs)))
        if dbs:
            fold_db.append(float(np.mean(dbs)))
        if aris:
            fold_ari.append(float(np.mean(aris)))

    rows_sel.append({
        'k': int(k),
        'folds_used': int(len(fold_eta2)),
        'eta2_wf_mean': float(np.nanmean(fold_eta2)) if fold_eta2 else np.nan,
        'eta2_wf_median': float(np.nanmedian(fold_eta2)) if fold_eta2 else np.nan,
        'kw_wf_mean': float(np.nanmean(fold_kw)) if fold_kw else np.nan,
        'sil_wf_mean': float(np.nanmean(fold_sil)) if fold_sil else np.nan,
        'ch_wf_mean': float(np.nanmean(fold_ch)) if fold_ch else np.nan,
        'db_wf_mean': float(np.nanmean(fold_db)) if fold_db else np.nan,
        'ari_wf_mean': float(np.nanmean(fold_ari)) if fold_ari else np.nan,
    })

SEL = pd.DataFrame(rows_sel)
SEL['z_eta'] = zcol(SEL['eta2_wf_mean'])
SEL['z_sil'] = zcol(SEL['sil_wf_mean'])
SEL['z_ch'] = zcol(SEL['ch_wf_mean'])
SEL['z_db'] = zcol(SEL['db_wf_mean'])
SEL['z_ari'] = zcol(SEL['ari_wf_mean'])
SEL['wf_score'] = 1.40*SEL['z_eta'] + 0.35*SEL['z_sil'] + 0.20*SEL['z_ch'] - 0.25*SEL['z_db'] + 0.20*SEL['z_ari']
SEL = SEL.sort_values('wf_score', ascending=False).reset_index(drop=True)
SEL.to_csv(MODEL_SEL_CSV, index=False)

BEST_K = int(SEL.iloc[0]['k'])
print('BEST_K =', BEST_K)
display(SEL)

# =========================
# Final fit train+val, predict all
# =========================
vec_final = TfidfVectorizer(
    analyzer='word',
    lowercase=False,
    token_pattern=r'[^\s]+',
    ngram_range=NGRAM_RANGE,
    min_df=MIN_DF,
    max_df=MAX_DF,
)

Xtv_sp = vec_final.fit_transform(texts[idx_tv])
Xall_sp = vec_final.transform(texts)

n_feat = Xtv_sp.shape[1]
if n_feat <= 2:
    Ztv = Xtv_sp.toarray()
    Zall = Xall_sp.toarray()
else:
    n_comp = int(min(SVD_DIM, max(2, n_feat - 1)))
    svd_final = TruncatedSVD(n_components=n_comp, random_state=SEED)
    Ztv = svd_final.fit_transform(Xtv_sp)
    Zall = svd_final.transform(Xall_sp)

sc_final = StandardScaler()
Ztv = sc_final.fit_transform(Ztv)
Zall = sc_final.transform(Zall)

KM = KMeans(n_clusters=BEST_K, n_init=N_INIT_KMEANS, random_state=SEED)
KM.fit(Ztv)
word_id = KM.predict(Zall)
word_dist = KM.transform(Zall).min(axis=1)

OUT = CTX[['ctx_window_id', 'base_window_id_end', 'start_ts', 'end_ts', 'split', 'token_seq', 'token_text', 'last_regime_id', 'ret_fwd_24h_pips']].copy()
OUT['word_id'] = word_id.astype(int)
OUT['word_token'] = OUT['word_id'].map(lambda x: f'w{x:02d}')
OUT['word_dist'] = word_dist.astype(float)

# Keep reduced contextual embedding for diagnostics
for j in range(min(Zall.shape[1], 64)):
    OUT[f'word_emb_{j:03d}'] = Zall[:, j]

# =========================
# Test profiling
# =========================
test_df = OUT[OUT['split'] == 'test'].copy()
prof = (test_df.groupby('word_id')['ret_fwd_24h_pips']
        .agg(n='count', mean_ret='mean', median_ret='median', std_ret='std')
        .reset_index())
prof['t_like'] = prof['mean_ret'] / prof['std_ret'].replace(0, np.nan)
prof = prof.sort_values('n', ascending=False).reset_index(drop=True)

# significance on test
kw_p = np.nan
anova_p = np.nan
eta2 = np.nan
groups = []
for _, g in test_df.groupby('word_id'):
    x = pd.to_numeric(g['ret_fwd_24h_pips'], errors='coerce').dropna().to_numpy()
    if len(x) >= MIN_CLUSTER_N_TEST:
        groups.append(x)

if len(groups) >= 2:
    kw_p = float(kruskal(*groups).pvalue)
    anova_p = float(f_oneway(*groups).pvalue)
    all_vals = np.concatenate(groups)
    gm = np.mean(all_vals)
    ssb = sum(len(g) * (np.mean(g) - gm) ** 2 for g in groups)
    sst = np.sum((all_vals - gm) ** 2)
    eta2 = float(ssb / sst) if sst > 0 else np.nan

# transition matrix between words
tr = pd.crosstab(OUT['word_token'], OUT['word_token'].shift(-1)).fillna(0)
tr_prob = tr.div(tr.sum(axis=1), axis=0).fillna(0.0)

# representative prototypes (closest to centroid)
proto_rows = []
for w in sorted(OUT['word_id'].unique()):
    d = OUT[OUT['word_id'] == w].copy().sort_values('word_dist').head(5)
    for _, r in d.iterrows():
        proto_rows.append({
            'word_id': int(w),
            'word_token': f'w{int(w):02d}',
            'ctx_window_id': int(r['ctx_window_id']),
            'start_ts': pd.Timestamp(r['start_ts']),
            'end_ts': pd.Timestamp(r['end_ts']),
            'word_dist': float(r['word_dist']),
            'token_text': str(r['token_text']),
        })
PROT = pd.DataFrame(proto_rows)

# save outputs
OUT.to_parquet(WORDS_PQ, index=False)
OUT.to_csv(WORDS_CSV, index=False)
prof.to_csv(PROFILE_CSV, index=False)
tr_prob.to_csv(TRANS_CSV)
PROT.to_csv(PROTOTYPES_CSV, index=False)

base_summary = {}
if IN_SUMMARY.exists():
    base_summary = json.loads(IN_SUMMARY.read_text())

summary = {
    'symbol': SYMBOL,
    'model_tag': 'phase6_context_words_v6',
    'source_model': 'phase5_v5_2_chronos_pretrained',
    'n_base_windows': int(len(base)),
    'n_context_windows': int(len(OUT)),
    'context_len': int(CONTEXT_LEN),
    'ngram_range': list(NGRAM_RANGE),
    'best_k_words': int(BEST_K),
    'split_counts': {k: int(v) for k, v in OUT['split'].value_counts().to_dict().items()},
    'kw_pvalue_test': None if not np.isfinite(kw_p) else float(kw_p),
    'anova_pvalue_test': None if not np.isfinite(anova_p) else float(anova_p),
    'eta2_test': None if not np.isfinite(eta2) else float(eta2),
    'wf_score_best_k': float(SEL.iloc[0]['wf_score']),
    'eta2_wf_mean_best_k': None if not np.isfinite(SEL.iloc[0]['eta2_wf_mean']) else float(SEL.iloc[0]['eta2_wf_mean']),
    'base_phase5_eta2_test': base_summary.get('eta2_test'),
    'base_phase5_kw_pvalue_test': base_summary.get('kw_pvalue_test'),
}
SUMMARY_JSON.write_text(json.dumps(summary, indent=2))

print('saved summary =', SUMMARY_JSON)
print('test words -> kw_p:', kw_p, '| anova_p:', anova_p, '| eta2:', eta2)
display(prof)
display(PROT.head(15))

# =========================
# Plots
# =========================
fig, axes = plt.subplots(1, 3, figsize=(18, 4.8))

axes[0].plot(SEL['k'], SEL['eta2_wf_mean'], marker='o', label='eta2_wf_mean')
ax2 = axes[0].twinx()
ax2.plot(SEL['k'], SEL['wf_score'], marker='s', color='tab:orange', label='wf_score')
axes[0].set_title('Phase6 K selection (words)')
axes[0].set_xlabel('k')
axes[0].set_ylabel('eta2_wf_mean')
ax2.set_ylabel('wf_score')

cnt = OUT['word_id'].value_counts().sort_index()
axes[1].bar(cnt.index.astype(str), cnt.values)
axes[1].set_title('Word counts (all context windows)')
axes[1].set_xlabel('word_id')
axes[1].set_ylabel('count')

emb_cols = [c for c in OUT.columns if c.startswith('word_emb_')]
E = OUT[emb_cols].to_numpy(dtype=float)
if E.shape[1] >= 2:
    xy = PCA(n_components=2, random_state=SEED).fit_transform(E)
    for w in sorted(OUT['word_id'].unique()):
        m = OUT['word_id'].to_numpy() == w
        axes[2].scatter(xy[m, 0], xy[m, 1], s=10, alpha=0.6, label=f'w{w:02d}')
    axes[2].legend(ncol=2, fontsize=8)
axes[2].set_title('PCA word-context embeddings')
axes[2].set_xlabel('PC1')
axes[2].set_ylabel('PC2')

plt.tight_layout()
plt.show()
```
